```python
import pandas as pd
import numpy as np
from statsmodels.stats.proportion import proportions_ztest
```

```python
df1 = pd.read_csv("open_ended_data_coded.csv").fillna(0)
df1
```

```python
df2 = pd.read_csv("Cleaned_study_data.csv")[['participant_code','treatment']]
df2
```

```python
df = df1.merge(df2, on="participant_code", how="right")
df
```

```python
treat = df[df.treatment == 1]
control = df[df.treatment == 0]
```

```python
D = []

for x in ['Open-minded','Popular','Activity','Civil','Uncivil','Extreme/Controversial',
          'Moderate','DK','No Answer','Non Sequitur','Hard to Code']:
    datum = {"code": x}
    t = treat[x]
    c = control[x]
    z = proportions_ztest([t.sum(), c.sum()],[t.size, c.size])
    D.append({"code": x,
              "control": c.mean(),
              "treat": t.mean(),
              "p": z[1],
              "z": z[0],
              "ratio": t.mean()/c.mean()})

pd.DataFrame(D).sort_values("ratio", ascending=False)
```
